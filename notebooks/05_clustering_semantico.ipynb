# 05 — Extra: clustering semántico (ligero)

Agrupamos descripciones de productos con TF-IDF + KMeans (sin modelos pesados).
En un FTE real se puede escalar a embeddings; aquí basta para mostrar la idea.

```python
import sys
from pathlib import Path
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
products = pd.read_csv(ROOT/'data'/'synthetic'/'products.csv')

vec = TfidfVectorizer(ngram_range=(1,2), min_df=1)
X = vec.fit_transform(products['descripcion'].fillna(''))
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
products['cluster'] = kmeans.fit_predict(X)
products.groupby('cluster')[['sku']].count().rename(columns={'sku':'n'}).join(
    products.groupby('cluster')['familia'].agg(lambda s: s.value_counts().index[0]).rename('familia_dominante')
)
```
